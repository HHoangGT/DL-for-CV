# Exploratory Data Analysis (EDA) - PASCAL VOC 2012 Segmentation

Notebook này nhằm mục đích hiển thị trực quan dữ liệu ảnh và mask từ tập dữ liệu PASCAL VOC 2012, chuẩn bị cho bài báo cáo.

```python
import sys
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
from tqdm.notebook import tqdm

sys.path.append("..")
from dataset.pascal_voc import VOCSegDataset, VOC_CLASSES
```

```python
# Khởi tạo dataset nhưng KHÔNG download nếu bạn đã tải trước đó.
# Đặt download=True nếu máy bạn chưa có dataset này.
dataset = VOCSegDataset(
    root="../data", year="2012", image_set="trainval", download=False
)
print(f"Tổng số lượng ảnh (trainval): {len(dataset)}")
```

```python
# Trực quan hoá ngẫu nhiên một vài samples
num_samples = 3
fig, axes = plt.subplots(num_samples, 2, figsize=(10, 5 * num_samples))

for i in range(num_samples):
    idx = np.random.randint(0, len(dataset))
    img, mask = dataset.get_raw(idx)

    axes[i, 0].imshow(img)
    axes[i, 0].set_title(f"Sample {idx} - Hình ảnh gốc")
    axes[i, 0].axis("off")

    axes[i, 1].imshow(mask, cmap="tab20", vmin=0, vmax=20)
    axes[i, 1].set_title(f"Sample {idx} - Dịch tễ (Ground Truth Mask)")
    axes[i, 1].axis("off")

plt.tight_layout()
plt.show()
```

```python
# Thống kê sơ bộ sự phân bố class (chạy rà soát trên 500 ảnh đầu tiên để demo)
class_counts = Counter()
for i in tqdm(range(min(500, len(dataset)))):
    _, mask = dataset.get_raw(i)
    mask_np = np.array(mask).flatten()
    class_counts.update(mask_np)

labels, counts = [], []
for cls_id in range(1, 21):  # Bỏ qua background (0) và ignore_index (255)
    labels.append(VOC_CLASSES[cls_id])
    counts.append(class_counts.get(cls_id, 0))

plt.figure(figsize=(12, 6))
plt.bar(labels, counts, color="skyblue")
plt.xticks(rotation=45, ha="right")
plt.title("Tần suất điểm ảnh (pixel) phân bổ theo từng lớp vật thể (trên 500 mẫu đầu)")
plt.ylabel("Số lượng lượng chênh pixel")
plt.tight_layout()
plt.show()
```
